# 2. Nettoyage des données

Contrôle des types, des valeurs manquantes, des doublons et des valeurs extrêmes, puis export du jeu de données nettoyé.

```python
import pandas as pd
data = pd.read_csv("../data/raw/winequality-red.csv")
```

```python
# Types des variables
data.dtypes
```

```python
# Les valeurs manquantes
data.isnull().sum()
```

```python
#Les doublons
data.duplicated().sum()
```

```python
# Extraction des valeurs
data.loc[data.duplicated(), :]
```

```python
# Suppression des doublons
data_clean = data.drop_duplicates()
```

```python
# Vérification
data_clean.duplicated().sum()
```

```python
# Détection d'outliers
data.describe()
```

```python
# Verification de dimension du tableau
print("Avant :", data.shape)
print("Après :", data_clean.shape)
```

```python
# Vérification global
data_clean.isnull().sum()
```

```python
# detection des doublons
data_clean.duplicated().sum()
```

```python
# Detection d'outliers
data_clean.describe()
```

```python
# Sauvegarder les données nettoyées
data_clean.to_csv("../data/processed/winequality_cleaned.csv", index=False)
print("Fichier nettoyé sauvegardé avec succès !")
```

**À retenir :** aucune valeur manquante, mais **240 doublons** supprimés. Le jeu nettoyé compte **1 359 vins** et sert de base à toutes les analyses suivantes. Les valeurs extrêmes (sucre résiduel, chlorures, sulfates) ont été conservées : elles sont plausibles chimiquement.
